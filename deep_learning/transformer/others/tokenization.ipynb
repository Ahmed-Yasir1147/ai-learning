````python
tokens = re.split(r"([.,:;!()\"'_?-]|\s)", corpus)
tokens = [token for token in tokens if not(token.strip() == "")]
tokens.extend(["<|unk|>", "<|eos|>"])
print(tokens)
````

````python
vocab = {token: index for index, token in enumerate(tokens)} 
````

````python
class SimpleTokenizer:
    def __init__(self, vocab):
        self.token_to_id = vocab
        self.id_to_token = {id: token for token, id in vocab.items()}

    def encode(self, text):
        tokens = re.split(r"([.,:;!()\"'_?-]|\s)", text) # convert to tokens
        tokens = [token for token in tokens if not(token.strip() == "")] # remove only space tokens
        tokens = [token if token in self.token_to_id else "<|unk|>" for token in tokens]
        return [self.token_to_id[token] for token in tokens] # convert to ids

    def decode(self, token_ids):
        tokens = [self.id_to_token[id] for id in token_ids] # conver to tokens
        text = " ".join(tokens) # combine to a sentence
        # return re.sub(r"\s+", text) # add spaces before & after every symbol
        return text
````

````python
tokenizer = SimpleTokenizer(vocab)
t1 = "He is good & funny."
t2 = "He is bad & hedious."
tokens = tokenizer.encode(t1 + " <|eos|> " + t2)
print("Tokens:", tokens)
corpus = tokenizer.decode(tokens)
print("Text:", corpus)
````

````python
from sentence_transformers import SentenceTransformer
import numpy as np
````

````python
embedder = SentenceTransformer("all-MiniLM-L6-v2")
````

````python
e1 = embedder.encode("I went to America and it was life changing experince")
e2 = embedder.encode("Machine learning is also software engineering")
e3 = embedder.encode("Travelling is fun")
````

````python
# Eucledian distance
print(np.linalg.norm(e1 - e2))
print(np.linalg.norm(e1 - e3))
````

````python
# Dot product
print(np.dot(e1, e2))
print(np.dot(e1, e3))
````

````python
# Cosine similarity
print(np.dot(e1, e2) / (np.linalg.norm(e1) * np.linalg.norm(e2)))
print(np.dot(e1, e3) / (np.linalg.norm(e1) * np.linalg.norm(e3)))
````
