### Streaming

```python
stream = client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Why rivers are non salty while oceans are salty?",
    stream=True
)

for event in stream:
    if event.event_type == "step.delta" and event.delta.type == "text":
        print(event.delta.text, end="", flush=True)
```

```python
stream = client.models.generate_content_stream(
    model="gemini-3.1-flash-lite",
    contents="Why rivers are non salty while oceans are salty?",
)

for chunk in stream:
    print(chunk.text, end="")
```

### Multi conversations

```python
interaction_one = client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="What is Machine learning?",
    generation_config={
        "max_output_tokens": 128,
    }
)
print("Bot:")
print(interaction_one.output_text)

interaction_two = client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Tell something about RAG",
    previous_interaction_id=interaction_one.id,
    generation_config={
        "max_output_tokens": 128,
    }
)
print("Bot:")
print(interaction_two.output_text)

interaction_three = client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Summarize what we have talked so far in this convo?",
    previous_interaction_id=interaction_two.id,
    generation_config={
        "max_output_tokens": 128,
    }
)
print("Bot:")
print(interaction_three.output_text)
```

### Structred output

```python
class AIScientist(BaseModel):
    name: str = Field(description="Full name",)
    birth_death: str = Field(description="Date of birth and death", )
    contributions: List[str] = Field(description="Major contributions to field of AI")
    controversies: List[str] = Field(description="Major controverises if any",)
```

```python
interaction = client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Adrew NJ",
    response_format={
        "type": "text",
        "mime_type": "application/json",
        "schema": AIScientist.model_json_schema()
    }
)

print(AIScientist.model_validate_json(interaction.output_text))
print(interaction.output_text)
```

### Function calling

```python
weather_function = {
    "type": "function",
    "name": "get_average_temperature",
    "description": "Get average temperature for given location.",
    "parameters": {
        "type": "object",
        "properties": {
            "location": {
                "type": "string",
                "description": "Cit name i.e, Lahore, Delhi, New York, etc",
            }
        },
        "required": ["location"]
    }
}
```

```python
interaction = client.interactions.create(
    model="gemini-3.1-flash-lite",
    input="Temperature of America",
    tools=[weather_function]
)
```

```python
for step in interaction.steps:
    if step.type == "function_call":
        print(f"Function to call: {step.name}")
        print(f"Arguments: {step.arguments}")
```

```python
print(interaction.output_text)
```

# Personal assistant

```python
class PersonalAssistant:

    def __init__(self, api_key=os.getenv("GEMINI_API_KEY_1")):
        self.client = genai.Client(api_key=api_key)
        print("Hi, I am your AI assistant. I am here to help you.")

    def ans_query(self, query, max_output_tokens=512):
        interaction = client.interactions.create(
            model="gemini-3.1-flash-lite",
            input=query,
            system_instruction="You are personal assistant. Be helpful.",
            generation_config={
                "temperature": 2.0,
                "max_output_tokens": max_output_tokens,

            },
        )
        return interaction.output_text.strip()

    def summarize_email(self, email, max_output_tokens=128):
        interaction = client.interactions.create(
            model="gemini-3.1-flash-lite",
            input=f"Summarize this email in two or three lines while for too rich email you may increase the output length: {email}",
            system_instruction="You are personal assistant. Be helpful.",
            generation_config={
                "temperature": 2.0,
                "max_output_tokens": max_output_tokens
            },
        )
        return interaction.output_text.strip()
```

```python
assistant = PersonalAssistant()
```

```python
query = input("What you want to know or talk about?")
response = assistant.ans_query(query)
print(response)
```

```python
email = input("Enter your email: ")
response = assistant.summarize_email(email)
print(response)
```
