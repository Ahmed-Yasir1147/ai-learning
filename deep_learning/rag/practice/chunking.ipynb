````python
text = """
Transformers are one of the most important developments in modern deep learning and natural language processing. The Transformer architecture was introduced to address limitations that existed in earlier sequence modeling approaches, particularly recurrent neural networks and long short-term memory networks. Instead of processing a sequence strictly from left to right, a Transformer can process many elements of a sequence in parallel while using attention mechanisms to determine which parts of the input are important for understanding each token.

The central idea behind the Transformer is the attention mechanism. Attention allows a model to consider relationships between different words or tokens in a sequence, even when those tokens are far apart. For example, in the sentence "The student placed the book on the table because it was heavy," understanding what "it" refers to requires considering information from earlier parts of the sentence. Self-attention gives the model a mechanism for establishing such relationships by comparing representations of different tokens.

A Transformer typically contains multiple layers, and each layer may contain a multi-head self-attention mechanism followed by a feed-forward neural network. Residual connections and layer normalization are also important components of the architecture. The input tokens are first converted into vector representations called embeddings. Because the Transformer does not inherently process tokens sequentially in the same way as a recurrent network, positional information must also be incorporated so that the model can distinguish between different positions in the sequence.

In self-attention, each input token is transformed into three representations known as a query, key, and value. The query represents what a token is looking for, while the key represents information that can be matched against queries. The value contains the information that will ultimately be aggregated. Attention scores are calculated by comparing queries with keys. These scores are scaled and passed through a softmax function to obtain weights. The weighted values are then combined to produce the output of the attention mechanism.

Multi-head attention extends this idea by allowing the model to perform several attention operations in parallel. Each attention head can potentially learn different relationships between tokens. One head might focus on syntactic relationships, another might focus on semantic relationships, and another might learn relationships involving positional or contextual information. The outputs of the different heads are concatenated and passed through another linear transformation.

The Transformer architecture can be divided into encoder and decoder components. The original Transformer proposed an encoder-decoder architecture for sequence-to-sequence tasks such as machine translation. The encoder receives the input sequence and creates contextual representations, while the decoder generates the output sequence one token at a time. The decoder uses masked self-attention so that it cannot access future tokens while generating the current token.

Encoder-only Transformer models are commonly used for tasks where understanding an input sequence is more important than generating new text. A well-known example is BERT, which uses bidirectional attention to create contextual representations of text. BERT can be fine-tuned for tasks such as text classification, named entity recognition, question answering, and semantic similarity. During pretraining, BERT learns from large amounts of text and develops general language representations that can later be adapted to specific tasks.

Decoder-only Transformer models are designed primarily for autoregressive text generation. GPT-style models belong to this category. During generation, the model predicts the next token based on the tokens that came before it. This process is repeated until the model produces a complete response. Large language models based on decoder-only Transformers can perform tasks such as question answering, summarization, code generation, reasoning, translation, and conversational interaction.

Encoder-decoder Transformers are particularly useful for tasks where one sequence needs to be transformed into another sequence. Models such as T5 use this approach by framing many natural language tasks as text-to-text problems. For example, summarization can be represented as an input containing a document and an output containing its summary. Translation can be represented as an English sentence being transformed into a sentence in another language.

Training a Transformer generally involves exposing the model to large amounts of data and optimizing its parameters using gradient-based optimization. The model calculates a loss that measures how different its predictions are from the expected outputs. Backpropagation is then used to calculate gradients, and an optimizer updates the model parameters. Modern Transformer training often involves very large datasets, distributed computing, specialized hardware, and techniques designed to make training more efficient.

One important advantage of Transformers is their ability to model long-range dependencies. Recurrent neural networks process sequences step by step, which can make it difficult to efficiently capture relationships between distant tokens. Transformers use attention to directly connect tokens with other tokens in the sequence. This makes it easier for the architecture to model relationships that may span large portions of the input.

However, standard self-attention also has a major limitation. Its computational and memory requirements increase significantly as the sequence length grows. If a sequence contains many tokens, the attention mechanism needs to calculate relationships between many pairs of tokens. This can make processing extremely long documents expensive. Researchers have therefore developed various approaches for efficient attention, sparse attention, sliding-window attention, and other techniques that attempt to reduce the computational cost of processing long contexts.

Tokenization is another important part of Transformer-based systems. Before text can be processed by a Transformer, it must be converted into tokens. A tokenizer divides text into smaller units that can be mapped to numerical identifiers. Depending on the tokenizer, a token might correspond to a complete word, part of a word, punctuation, or another textual unit. The resulting token IDs are passed to the model, which converts them into embeddings before processing them through the Transformer layers.

The size of the context window is particularly important when working with Transformer models. A model cannot necessarily process an arbitrarily large document in a single request. If a document contains more tokens than the model's context window allows, the document must be divided into smaller pieces. This is one reason document chunking is important in retrieval-augmented generation systems.

In a retrieval-augmented generation system, documents are usually loaded, cleaned, split into chunks, embedded, and stored in a vector database. When a user asks a question, the question is converted into an embedding and compared against the stored document embeddings. The most relevant chunks are retrieved and supplied to the language model as context. The Transformer can then use the retrieved information to generate an answer.

Chunk size has a significant effect on the performance of a retrieval system. If chunks are extremely small, important information may be separated across multiple chunks. The retriever may find one part of an explanation while missing the surrounding context. On the other hand, if chunks are extremely large, each retrieved chunk may contain a large amount of irrelevant information. Larger chunks can also reduce the number of distinct pieces of information that fit into the model's context window.

Chunk overlap is often used to reduce the problem of information being split at an unfortunate boundary. For example, suppose a paragraph is divided into two chunks and the second chunk begins halfway through an explanation. A small overlap means that some text from the end of the first chunk is repeated at the beginning of the second chunk. This can help preserve context across chunk boundaries, although excessive overlap increases storage requirements and can lead to redundant retrieval results.

The choice of splitting strategy is also important. A basic character-based splitter treats the document primarily as a sequence of characters and uses separators such as spaces, newlines, or punctuation to determine possible boundaries. A recursive character splitter can try several separators in order. It may first attempt to split at paragraph boundaries, then line boundaries, then sentence-like boundaries, and finally individual characters if necessary.

Structure-aware splitting can be more appropriate for structured documents. For example, a Markdown document may contain headings, subheadings, paragraphs, lists, and code blocks. A Markdown-aware splitter can use these structures to create chunks that preserve relationships between headings and their corresponding content. Similarly, a programming-language-aware splitter can attempt to preserve functions, classes, and other logical code structures rather than arbitrarily dividing source code based only on character positions.

Transformers have also had a major impact beyond natural language processing. The same general architectural principles have been adapted for computer vision, speech processing, multimodal learning, biological sequence modeling, and other domains. Vision Transformers, for example, divide images into patches and process those patches as a sequence of representations. Multimodal Transformer systems can combine information from text, images, audio, and other types of data.

The success of Transformers has also led to the development of increasingly large foundation models. These models are trained on broad datasets before being adapted for particular applications. Pretraining allows a model to learn general patterns from data, while fine-tuning or instruction tuning can make the model more useful for specific tasks. Additional techniques such as reinforcement learning from human feedback, preference optimization, retrieval augmentation, and tool use can further modify how a model behaves.

Despite their capabilities, Transformer models do not automatically guarantee factual accuracy. A model can generate plausible statements that are incorrect or unsupported by evidence. Retrieval-augmented generation attempts to address some of these problems by providing external information to the model at inference time. However, the quality of the retrieved information is critical. If the document splitter creates poor chunks, the embedding model may represent those chunks poorly, the retriever may return irrelevant information, and the final Transformer may generate an answer based on incomplete context.

For this reason, document preprocessing is an important part of a practical RAG pipeline. Splitting should not be considered merely a technical step for satisfying a model's context limit. The goal is to create chunks that represent coherent units of information. Good chunks should ideally contain enough context to answer a question while remaining focused enough that their embeddings accurately represent their main topic.

Ultimately, Transformers provide the underlying architecture that allows modern language models to understand and generate sequences using attention-based representations. Document splitting operates one level outside the Transformer itself. It prepares large external documents so that relevant pieces can be efficiently retrieved and placed into the model's context window. Understanding the relationship between tokenization, chunking, embeddings, retrieval, context windows, and Transformer processing is therefore essential when designing a modern retrieval-augmented generation system."""
````

````python
markdown = """
# Thermometer
A simple and customizable widget to be used as a thermometer in Flutter.

## Features

This widget will create a thermometer. When widget is created inside a stateful widget and temperature is changed in set state method, then the liquid of thermometer animates from old to new temperature.

## Getting started

- [Installation](#installation)
- [Constructor](#constructor)
- [Usage_Example](#usage-example)


### Installation

Add the dependency to your `pubspec.yaml`:

```yaml
dependencies:
  thermometer_widget: ^0.1.0
```

And then run following command on terminal:

```bash
flutter pub get
```

Add the following import statement in the dart file you are going to use this widget:

```dart
import 'package:thermometer/thermometer.dart';
```

### Constructor

| Property | Default | Description |
|-----------|----------|-------------|
| height |  | The height of thermometer. Width will automatically get 25% of it. |
| duration | const Duration(milliseconds: 300) | The time duration of animation of liquid when temperature is changed. |
| temperatureC |  | The temperature of thermometer in Celsius. |
| showScreenTemperatureInFahrenheit | false | To show the temperature on screen in Fahrenheit. |
| backgroundGradient| const LinearGradient(colors: [Colors.white, Color.fromARGB(255, 224, 224, 224)], stops: [0.3, 1.0], begin: Alignment.centerLeft, end: Alignment.centerRight,) | Gradient which will act as background (color) of thermometer. |
| textColor | const Color.fromARGB(255, 87, 87, 87) | Text color used inside the thermometer. |
| boundaryColor | const Color.fromARGB(255, 189, 189, 189) | Color of the thermometer boundary. |
| screenColor | const Color.fromARGB(255, 224, 224, 224) | Background color of the screen area. |
| tubeColor | Colors.white | Color of the thermometer tube. |
| screenTextColor | const Color.fromARGB(255, 66, 66, 66) | Color of the temperature text displayed on the screen. |
| readingLinesColor | Colors.black | Color of the small reading lines inside the thermometer. |
| veryFreezingTemperatureColor | const Color.fromARGB(255, 10, 66, 122) | Liquid color for very low (freezing) temperature range between -20°C and lesser temperature.|
| freezingTemperatureColor | const Color.fromARGB(255, 31, 135, 221) | Liquid color for low temperature range between -21°C and 0°C. |
| normalTemperatureColor | const Color.fromARGB(255, 50, 167, 54) | Liquid color for normal temperature range between 1°C and 10°C. |
| hotTemperatureColor | const Color.fromARGB(255, 219, 200, 22) | Liquid color for high temperature range between 11°C and 30°C. |
| veryHotTemperatureColor | const Color.fromARGB(255, 189, 32, 21) | Liquid color for very high (hot) temperature range between 31°C and greater temperature |

### Usage Examples

#### Example 
Here is a basic example.
It consists of thermometer widget and a button. On clicking button, the value of temperature changes and liquid animates from last to new temperature.

```dart
import 'package:flutter/material.dart';
import 'package:thermometer/thermometer.dart';

void main() {
  runApp(const MyApp());
}

class MyApp extends StatelessWidget {
  const MyApp({super.key});

  @override
  Widget build(BuildContext context) {
    return MaterialApp(
      home: const MyHomePage(),
      debugShowCheckedModeBanner: false,
    );
  }
}

class MyHomePage extends StatefulWidget {
  const MyHomePage({super.key});

  @override
  State<MyHomePage> createState() => _MyHomePageState();
}

class _MyHomePageState extends State<MyHomePage> {
  double temp = -10.39;
  int counter = 1;

  List<double> temps = [-10.39, 40, 4.5, -39.33, 25.44,];

  void _changeTemperature() {
    setState(() {
      temp = temps[counter];
      counter++;
      if (counter == 5) {
        counter = 0;
      }
    });
  }

  @override
  Widget build(BuildContext context) {
    final Size(:height, :width) = MediaQuery.of(context).size;
    return Scaffold(
      appBar: AppBar(
        backgroundColor: Colors.grey,
        title: Text(
          'Custom Thermometer',
          style: TextStyle(color: Colors.white),
        ),
      ),
      body: Center(
        child: Container(
          padding: EdgeInsets.only(bottom: height * 0.1),
          child: Thermometer(
            height: 400,
            temperatureC: temp,
          ),
        ),
      ),
      floatingActionButton: FloatingActionButton(
        backgroundColor: Colors.grey.shade50,
        onPressed: _changeTemperature,
        tooltip: 'Change Temperature',
        child: Icon(Icons.star, color: Colors.grey.shade800),
      ),
    );
  }
}
```

![Thermometer example](https://github.com/Ahmed-Yasir1147/Thermometer/blob/main/assets/images/example_1.gif?raw=true) """
markdown_doc = Document(page_content=markdown)
````

# Agentic chunking

````python
from pydantic import BaseModel, Field
from typing import List
from transformers import pipeline
````

````python
pipe = pipeline("text-generation", model="HuggingFaceTB/SmolLM2-135M-Instruct")
````

````python
prompt = f"""
    You are an expert document analyzer. Your task is to perform agentic chunking on the provided text.
    Read through the text carefully and segment it into smaller, logically complete, and self-contained 
    blocks of information (chunks). Do not split mid-thought or separate closely related context; 
    instead, break at natural topic or structural transitions and each chunk shouldn't be single worded it should have complete thought.
    Each chunk should be separated by two escape sequences
    Text to chunk:
    {text}
"""
````

````python
output = pipe(prompt)
````

````python
extracted_output = output[0]["generated_text"]
chunks = extracted_output.split("\n\n")
````

````python
for i, chunk in enumerate(chunks):
    print(i, " : ", chunk)
    
````
