To extract text from image in pdf,

1. First load and save images from pdf
2. Extract text from saved images using OCR

```python
import pymupdf
```

```python
pdf = pymupdf.open("dataset/java_outline.pdf")

page = pdf[0]

images = page.get_images()

for index, image in enumerate(images):
    xref = image[0]

    pix = pymupdf.Pixmap(pdf, xref)
    pix.save(f"dataset/java_outline_{index}.png")
```
