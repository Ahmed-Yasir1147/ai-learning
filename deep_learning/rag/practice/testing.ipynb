```python
import outlines
from pydantic import BaseModel, Field
from transformers import AutoModelForCausalLM, AutoTokenizer
from typing import List
```

```python
model = outlines.from_transformers(
    AutoModelForCausalLM.from_pretrained("occ-ai/OCC-RAG-0.6B", device_map="auto"),
    AutoTokenizer.from_pretrained("occ-ai/OCC-RAG-0.6B")
)
```

```python
class MultiQueries(BaseModel):
    queries: List[str] = Field(description="List of queries")
```

```python
response = model(
    "Give three versions of this query: Define transformers in AI with examples.",
    MultiQueries
)
```
