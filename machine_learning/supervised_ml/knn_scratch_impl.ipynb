```python
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error
```

```python
class KNNClassifier:

    def __init__(self, k=3):
        self.k=k

    def fit(self, X, y):
        # Stores data
        self.X_train = X
        self.y_train = y

    def _eculedian_distance(self, x1, x2):
        return np.sqrt(np.sum((x1 - x2) ** 2))

    def _predict_one(self, x):
        # Calucate distnace
        distances = [self._eculedian_distance(x, x_train) for x_train in self.X_train]

        # Find nearest k neighbors
        # argsort() sorts the elements and return array of their original distance
        sorted_indices = np.argsort(distances)
        knn_indices = sorted_indices[:self.k]
        # output classes at each index
        knn_classes = [self.y_train[i] for i in knn_indices]

        # Find majority class
        # bincount() returns list containing frequency of each category where index represents category
        return np.argmax(np.bincount(knn_classes))

    def predict(self, X):
        return [self._predict_one(x) for x in X]
        
            
```

```python
X = np.array([[1,2], [3,4], [5,6], [7,8]])
y = np.array([0, 0, 0, 1])

model = KNNClassifier();
model.fit(X, y)

print(y)
print(model.predict([[10,20]]))
```

```python
class KNNRegressor:

    def __init__(self, k=3):
        self.k=k

    def fit(self, X, y):
        # Stores data
        self.X_train = X
        self.y_train = y

    def _eculedian_distance(self, x1, x2):
        return np.sqrt(np.sum((x1 - x2) ** 2))

    def _predict_one(self, x):
        # Calucate distnace
        distances = [self._eculedian_distance(x, x_train) for x_train in self.X_train]

        # Find nearest k neighbors
        # argsort() sorts the elements and return array of their original distance
        sorted_indices = np.argsort(distances)
        knn_indices = sorted_indices[:self.k]
        # output classes at each index
        knn_outputs = [self.y_train[i] for i in knn_indices]

        # Find majority output
        return np.mean(knn_outputs)

    def predict(self, X):
        return [self._predict_one(x) for x in X]
        
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = X.flatten() * 2

model = KNNRegressor()
model.fit(X, y)

y_test = [[1.5], [3.5], [6]]

y_pred = model.predict(y_test)

print(y_test)
print(y_pred)
```
