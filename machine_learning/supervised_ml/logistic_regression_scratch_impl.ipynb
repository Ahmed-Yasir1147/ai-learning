```python
import numpy as np
```

```python
class LogisticRegressor:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.coefficients = None
        self.learning_rate = learning_rate
        self.n_iter = n_iter

    def _sigmoid(self, z):
        return (1 / (1 + np.exp(-z))) 

    def fit(self, X, y):
        m, n = X.shape

        # Intializing bias and coefficients
        self.bias = 0
        self.coefficients = np.zeros(n)

        # Gradient descent
        for i in range(self.n_iter):
            # Predict using hypothesis function
            z = self.bias + np.dot(X, self.coefficients)
            # Passing linear regression result to sigmoid function
            y_pred = self._sigmoid(z)

            # Derivatives of cost function with respect to bias and coefficients
            db = (1 / m) * np.sum(y_pred - y)
            dc = (1 / m) * np.dot(X.T, (y_pred - y))

            # Graident updation of bias and coefficients
            self.bias -= self.learning_rate * db
            self.coefficients -= self.learning_rate * dc

    def get_prob(self, X):
        z = self.bias + np.dot(X, self.coefficients)
        return self._sigmoid(z)
        

    def predict(self, X, threshold=0.5):
        prob = self.get_prob(X)
        y_pred = prob >= threshold
        return y_pred.astype(int)
        
```

```python
X = np.array([[1, 2], [3, 4], [5, 6], [7, 8]])
y = np.array([0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

print(model.predict([[4, 5]]))

print(y)
```
