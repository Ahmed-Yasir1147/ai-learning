```python
import numpy as np
import pandas as pd
from sklearn.metrics import r2_score, mean_squared_error
```

```python
class LinearRegression:

    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.coefficients = None
        self.learning_rate = learning_rate
        self.n_iter = n_iter

    def fit(self, X, y):
        m, n = X.shape
        # Initialization of bias and coefficient
        self.bias = 0
        self.coefficients = np.zeros(n)

        # Gradient descent
        for i in range(self.n_iter):
        
            # Hypothesis function
            y_pred = self.bias + np.dot(X, self.coefficients)

            # Derivative with respect to bias and coeffeicients
            db = (1 / m) * np.sum(y_pred - y)
            dc = (1 / m) * np.dot(X.T, (y_pred - y))

            # Gradient 
            self.bias = self.bias - self.learning_rate * db
            self.coefficients = self.coefficients - self.learning_rate * dc
        
        
    def predict(self, X):
        return self.bias + np.dot(X, self.coefficients) 
        
```

```python
class LinearRegressionOLS:

    def __init__(self):
        self.bias = None
        self.coefficients = None

    def fit(self, X, y):
        m, n = X.shape
        X_updated = np.c_[np.ones(m), X]

        result = np.linalg.inv(X_updated.T @ X_updated) @ (X_updated.T @ y)
        self.bias = result[0]
        self.coefficients = result[1:]
        
        
    def predict(self, X):
        return self.bias + np.dot(X, self.coefficients) 
```

```python
# Testing
X = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])
y = X.flatten() ** 2

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)


print("R2:", r2_score(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))
```

```python
# Testing OLS
X = np.array([[1], [2], [3], [4], [5], [6], [7], [8]])
y = X.flatten() ** 2

model = LinearRegressionOLS()
model.fit(X, y)

y_pred = model.predict(X)

print("R2:", r2_score(y, y_pred))
print("MSE:", mean_squared_error(y, y_pred))
```
