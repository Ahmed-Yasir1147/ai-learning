```python
import gymnasium as gym
import numpy as np
import random
```

```python
# create cliff walking environment
env = gym.make("CliffWalking-v1")
```

```python
# environment properties
print(f"Number of states: {env.observation_space.n}")
print(f"Number of actions: {env.action_space.n}")
starting_point, _ = env.reset() 
print(f"Starting point: {starting_point}") # (3, 0) => 3 * 12 + 0 = 36
```

# Training

```python
# Hyper parameters
gamma = 0.99
alpha = 0.5
episodes = 500
epsilon = 0.1
```

```python
# Q table
Q = np.zeros((48, 4))
```

```python
# Policy: Epsilon - Greedy
def epsilon_greedy(state):
    if random.random() < epsilon:
        # Exploration
        return env.action_space.sample() # choosing random action
    else:
        # Exploitation
        return np.argmax(Q[state]) # choosing index(action) with highest reward for given state
```

```python
for episode in range(episodes):

    done = False # variable to store if episode is completed

    state, _ = env.reset() # resetting environment for each episode
     # first action

    total_reward = 0 # total reward during episode
    episode_len = 0 # number of steps taken during episode

    while not done:
        action = epsilon_greedy(state)
        next_state, reward, terminated, truncated, _ = env.step(action)
        next_action = epsilon_greedy(next_state)
        done = terminated or truncated

        # SARSA updated
        Q[state, action] += alpha * (reward + gamma * (Q[next_state, next_action] - Q[state, action])) 

        state = next_state

        total_reward += reward
        episode_len += 1

    
    print(f"Episode: {episode} => Reward: {total_reward} , Length: {episode_len}")

env.close()
```

```python
# Agent Learning
env = gym.make("CliffWalking-v1", render_mode="human")
state, _ = env.reset()
done = False
total_reward = 0
episode_len = 0

while not done:
    action = np.argmax(Q[state])
    next_state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated
    next_action = epsilon_greedy(next_state)

    state = next_state
    total_reward += reward
    episode_len += 1

print(f"Reward: {total_reward} Length: {episode_len}")
env.close()
```
