```python
import gymnasium as gym
import numpy as np
import random
```

```python
env = gym.make("CliffWalking-v1")
```

```python
alpha = 0.4
gamma = 0.9
episodes = 500
epsilon = 0.1
```

```python
Q = np.zeros((48, 4))
```

```python
def epsilon_greedy(state):
    if random.random() < 0.1:
        return env.action_space.sample()
    else:
        return np.argmax(Q[state])
```

```python
for episode in range(episodes):

    done = False
    state, _ = env.reset()

    total_reward = 0
    episode_len = 0

    while not done:
        action = epsilon_greedy(state)
        next_state, reward, terminated, turncated, _ = env.step(action)
        done = terminated or turncated

        # Q learning update
        Q[state, action] += alpha * (reward + gamma * (np.max(Q[next_state]) - Q[state, action]))

        state = next_state
        total_reward += reward
        episode_len += 1

    print(f"Episode: {episode} => Reward: {total_reward} , Length: {episode_len}")

env.close()
```

```python
env = gym.make("CliffWalking-v1", render_mode="human")
total_reward = 0
episode_len = 0
state, _ = env.reset()
done = False

while not done:
    action = np.argmax(Q[state])
    next_state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated

    total_reward += reward
    episode_len += 1
    state = next_state

env.close()
```
