```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
from sklearn.cluster import DBSCAN
```

```python
iris_df = load_iris()

X = iris_df.data
y = iris_df.target
```

```python
# scaling
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
fig, ((axis1, axis2) , (axis3, axis4)) = plt.subplots(2, 2, figsize=(10, 10))
sns.scatterplot(
    x=X[:, 0],
    y=X[:, 1],
    ax=axis1,
    c=y
)
sns.scatterplot(
    x=X[:, 2],
    y=X[:, 3],
    ax=axis2,
    c=y
)
sns.scatterplot(
    x=X[:, 0],
    y=X[:, 2],
    ax=axis3,
    c=y
)
sns.scatterplot(
    x=X[:, 1],
    y=X[:, 3],
    ax=axis4,
    c=y
)
```

```python
dbscan = DBSCAN(
    eps=0.8,
    min_samples=5,
)
labels = dbscan.fit_predict(X)
```

```python
sns.scatterplot(
    x=X[:, 1],
    y=X[:, 2],
    c=labels
)
```
